```python
#Write a Progrm to find the factors of a number
num = int(input("Enter the number"))
fact = []
for i in range(1,num +1):
  if num % i == 0:
    fact.append(i)
print("Factors are :", fact)
```

```python
#Write a Progarm in Python to Check the frequency of a word in a sentence

sentence = "Hello Human you are such a lovely human human are such lovely human human human human human"
l1 = sentence.split()
wordFrequency = {}

for word in l1:
  if word in wordFrequency.keys():
    wordFrequency[word] += 1
  else:
    wordFrequency[word] = 1
print(wordFrequency)
```
